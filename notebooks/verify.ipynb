## Verifying the cluster data

```python
import pandas as pd
import numpy as np
```

### Load data

```python
data = pd.read_csv('../data/data_cleaned_titles.csv')
locations = data[["id", "lat", "long"]]
clusters = pd.read_csv('../data/predictions.csv')
kmeans_clusters = clusters[["id", "kmean"]]
dbscan_clusters = clusters[["id", "dbscan"]]
hdbscan_clusters = clusters[["id", "hdbscan"]]
```

```python
# Verify if all points belong to a single cluster
kmeans_unique_clusters = kmeans_clusters['kmean'].nunique()
dbscan_unique_clusters = dbscan_clusters['dbscan'].nunique()
hdbscan_unique_clusters = hdbscan_clusters['hdbscan'].nunique()

print(f"KMeans unique clusters: {kmeans_unique_clusters}")
print(f"DBSCAN unique clusters: {dbscan_unique_clusters}")
print(f"HDBSCAN unique clusters: {hdbscan_unique_clusters}")
```

```python
# Verify that the clusters' convex hulls don't overlap
```

```python
# Generate a CSV with the id, latitude, longitude, cluster for each algorithm
merged_kmeans = pd.merge(locations, kmeans_clusters, on='id')
merged_dbscan = pd.merge(locations, dbscan_clusters, on='id')
merged_hdbscan = pd.merge(locations, hdbscan_clusters, on='id')
total_merged = pd.merge(merged_kmeans, merged_dbscan, on=['id', 'lat', 'long'])
total_merged = pd.merge(total_merged, merged_hdbscan, on=['id', 'lat', 'long'])
total_merged.to_csv('../data/verification.csv', index=False)
```
